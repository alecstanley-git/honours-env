# Blackbody spectra

## Planck Equations 

Flux density (flux per unit frequency) per unit soild angle 

\begin{equation}
B_\nu(\nu, T) = \frac{2 h \nu^3}{c^2} \frac{1}{{\rm exp}(h\nu/kT) - 1}
\end{equation}

Flux per unit wavelength per unit solid angle 

\begin{equation}
B_\lambda(\lambda, T) = \frac{2 h c^2}{\lambda^5} \frac{1}{{\rm exp}(hc/\lambda kT) - 1}
\end{equation}



```python
# Function for the blackbody curve 
# Inputs are frequency and temperature in SI units (Hz, Kelvin)
# Output is flux density for a one steradian object in SI units (W/m2/Hz/Steradian)

def Bnu(nu, T):
    
    Bnu = 2.0*scipy.constants.h*nu**3.0/scipy.constants.c**2 * 1.0/(np.exp(scipy.constants.h*nu/(scipy.constants.k*T))-1.0)
    return Bnu

# Function for the blackbody curve 
# Inputs are wavelength and temperature in SI units (m, Kelvin)
# Output is flux density for a one steradian object in SI units (W/m2/m/Steradian)

def Blam(lam, T):
    
    Blam = 2.0*scipy.constants.h*scipy.constants.c**2.0/lam**5 * 1.0/(np.exp(scipy.constants.h*scipy.constants.c/(lam*scipy.constants.k*T))-1.0)
    return Blam
```
