```python
# Importing Libraries

import numpy as np
import astropy
import photutils
import ccdproc
from ccdproc import CCDData, combiner
from astropy import units as u
import astropy.io.fits as fits
from astropy.io import ascii

from astropy.visualization import SqrtStretch
from astropy.visualization.mpl_normalize import ImageNormalize
from astropy.wcs import WCS
from astropy.coordinates import SkyCoord
import matplotlib.pyplot as plt
from matplotlib.colors import LogNorm
from photutils.centroids import centroid_com, centroid_1dg, centroid_2dg
from photutils.aperture import CircularAperture
from photutils.aperture import aperture_photometry
from photutils.background import Background2D
from photutils.background import MedianBackground
from photutils.detection import DAOStarFinder
from photutils.segmentation import detect_sources, deblend_sources, SourceCatalog
from scipy.ndimage import shift
import astroalign as aa
import gc

from astropy.coordinates import SkyCoord
```

```python
# Path for aligned images
path = "Alligned_S50_faeae72a_20260321_debayer/NGC_2175_sub/"
r_images = ccdproc.ImageFileCollection(path, glob_include="*r.fit")
g_images = ccdproc.ImageFileCollection(path, glob_include="*g.fit")
b_images = ccdproc.ImageFileCollection(path, glob_include="*b.fit")
```

```python
# Reading and storing scientific images

r_scim = []
r_names = []
g_scim = []
g_names = []
b_scim = []
b_names = []
for fn in r_images.files:
    r_scim.append(
        CCDData.read(path + fn, unit="adu")
    )  # Reading data and adding the image to scim list
    r_names.append(fn)

del r_images
gc.collect()

for fn in g_images.files:
    g_scim.append(
        CCDData.read(path + fn, unit="adu")
    )  # Reading data and adding the image to scim list
    g_names.append(fn)

del g_images
gc.collect()

for fn in b_images.files:
    b_scim.append(
        CCDData.read(path + fn, unit="adu")
    )  # Reading data and adding the image to scim list
    b_names.append(fn)

del b_images
gc.collect()
```
